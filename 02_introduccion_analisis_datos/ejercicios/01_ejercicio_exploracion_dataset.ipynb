### 4. Datos faltantes

```python
# ¿Tiene datos faltantes el dataset?
# Suma de nulos detectados por pandas tras la lectura inicial
print('Nulos detectados inicialmente:')
print(df.isnull().sum())
print('\nNulos específicos en workclass/occupation:')
print(df[['workclass', 'occupation']].isnull().sum())
# Detectar columnas que contienen '?' (con o sin espacios) -- sin depender de conocer las columnas
mask_cols = df.apply(lambda col: col.astype(str).str.strip().eq('?').any())
print('\nColumnas con "?":', mask_cols[mask_cols].index.tolist())
# Contar cuántos '?' por columna
counts_q = df.apply(lambda col: col.astype(str).str.strip().eq('?').sum())
print('\nConteo de "?" por columna (antes de reemplazar):')
print(counts_q[counts_q > 0])
# Reemplazar cualquier celda que sea solo '?' (con o sin espacios) por NA y luego comprobar de nuevo
df = df.replace(r'^\\s*\\?\\s*$', pd.NA, regex=True)
print('\nNulos después de normalizar "?" -> NA:')
print(df.isna().sum())
# Buscar columnas que contengan '?' en cualquier posición (si apareciera en textos largos)
contains_q = df.apply(lambda c: c.astype(str).str.contains(r'\\?', regex=True).any())
print('\nColumnas que contienen "?" en cualquier posición:', contains_q[contains_q].index.tolist())
```

**Respuesta:** Sí. En este dataset los faltantes están representados por `?` en `workclass` y `occupation`. Con `na_values=["?"]` y `skipinitialspace=True` se detectan 1.836 faltantes en `workclass` y 1.843 en `occupation`; el resto no tiene nulos.
